### Problem 4

```python
A = 2e11 #s-1
E = 65e3 #J/mol

T = np.array([300, 350, 400, 450])
k = A * np.exp(-1*E/(8.314*T))
print(k)

print(k[2]/k[0])
```

### Problem 5

```python
rho = np.array([812.4, 813.1, 811.8, 812.9, 812.2, 813.4])
mean = rho.mean()
std = np.std(rho, ddof=1)
print(rho)
print(f'mean = {mean}')
print(f'std deviation = {std}')

#to find furthest from mean value
farthest = rho[np.argmax(np.abs(rho - mean))]
print(farthest)
```

### Problem 6

```python
Re = np.array([1500, 5200, 800, 96000, 3500, 42000, 2100])
mask = Re>4000
count = mask.sum()
turbulent = Re[mask]

print(f'count = {count} : turbulent  = {turbulent}')
```

### Problem 7

```python
readings = np.array([[420.1, 421.5, 419.8], [455.3, 454.9, 455.8], [438.6, 439.2, 438.1]])

mean_row = np.mean(readings, axis=1)
mean_column = np.mean(readings, axis=0)
print(f'axis=1 -> row_avg = {mean_row} : axis=0 -> column_avg = {mean_column}')
```

### Problem 8

```python
c, a = np.loadtxt('data/calibration.csv', delimiter=',', skiprows=1, unpack=True)
print(a.size)
print(np.max(a))

fig, ax = plt.subplots(figsize=(7,4.5))
ax.set_xlabel("Concentration (mMol/L)")
ax.set_ylabel("Absorbance")
ax.scatter(c, a, color="green", marker="o", label="experimental")
x = np.linspace(0, 10, 11)
y = 0.0521*x
ax.plot(x, y, label="theoretical")
ax.set_title("Plot of Concentration v/s Absorbance")
ax.legend()
plt.show()
```
