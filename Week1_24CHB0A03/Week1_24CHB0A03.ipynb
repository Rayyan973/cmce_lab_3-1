### Problem 9: mixture_MW function

**Pseudocode:**
1. Define mixture_MW(y, MW) that loops over keys in y
2. Accumulate sum of y[i] * MW[i]
3. Return the mixture MW
4. Apply to flue gas composition and assert pure N2 check

```python
def mixture_MW(y, MW):
    """
    Compute mixture molecular weight from mole fractions and component MWs.
    y  : dict of mole fractions [-]
    MW : dict of molecular weights [g/mol]
    returns mixture MW [g/mol]
    """
    M = 0.0
    for comp in y:
        M += y[comp] * MW[comp]
    return M

MW = {"CO2": 44.01, "H2O": 18.015, "N2": 28.02, "O2": 32.00} #g/mol
y = {"CO2": 0.12, "H2O": 0.07, "N2": 0.74, "O2": 0.07} #-

print(f'flue gas M = {mixture_MW(y, MW):.3f} g/mol')
assert mixture_MW({"N2": 1.0}, MW) == 28.02
print('assert passed: pure N2 returns 28.02')
```

### Problem 10: Psat table

**Pseudocode:**
1. Store Antoine constants for ethanol and acetone
2. For each T in [20, 40, 60]:
   - compute Psat = 10^(A - B/(C+T)) for both species
   - print aligned row
3. Compare columns: higher Psat => more volatile

```python
import math

antoine = {
    "ethanol": (8.20417, 1642.89, 230.30),
    "acetone": (7.02447, 1161.00, 224.00),
}

temps = [20, 40, 60] #C

print(f'{"T(C)":>6}  {"ethanol Psat(mmHg)":>20}  {"acetone Psat(mmHg)":>20}')
for T in temps:
    row = []
    for sp in ["ethanol", "acetone"]:
        A, B, C = antoine[sp]
        Psat = 10 ** (A - B / (C + T)) #mmHg
        row.append(Psat)
    print(f'{T:6d}  {row[0]:20.2f}  {row[1]:20.2f}')

print('acetone is more volatile at every temperature (higher Psat)')
```

### Problem 11: successive substitution

**Pseudocode:**
1. Set x_old = 0.5, tol = 1e-10, max_iter = 200
2. Loop: x_new = exp(-x_old), dx = |x_new - x_old|
3. Print first 5 traces
4. Stop when dx < tol or hit max_iter
5. Report x* and iteration count

```python
x_old = 0.5
tol = 1e-10
max_iter = 200

for i in range(1, max_iter + 1):
    x_new = math.exp(-x_old)
    dx = abs(x_new - x_old)
    if i <= 5:
        print(f'iter {i}: x = {x_new:.10f}, |dx| = {dx:.3e}')
    if dx < tol:
        print(f'converged: x* = {x_new}, iterations = {i}')
        break
    x_old = x_new
else:
    print(f'did not converge in {max_iter} iterations')
```

### Problem 12: flow_regime

**Pseudocode:**
1. Compute Re = rho*v*D / mu
2. If Re < 2100 -> laminar; elif Re <= 4000 -> transition; else turbulent
3. Return (Re, regime) tuple
4. Call once with keyword args in a different order

```python
def flow_regime(rho, v, D, mu):
    """
    Classify pipe flow regime from Reynolds number.
    rho : density [kg/m3]
    v   : velocity [m/s]
    D   : diameter [m]
    mu  : viscosity [Pa.s = kg/(m.s)]
    returns (Re [-], regime [str])
    """
    Re = rho * v * D / mu
    if Re < 2100:
        regime = "laminar"
    elif Re <= 4000:
        regime = "transition"
    else:
        regime = "turbulent"
    return (Re, regime)

# keyword args in different order from definition
print(flow_regime(mu=0.001, D=0.05, v=1.5, rho=998))
```

### Problem 13: friction_factor

**Pseudocode:**
1. If Re < 2100: f = 64/Re (laminar)
2. Else: f = 0.316 * Re^(-0.25) (Blasius)
3. Assert laminar check and verify Blasius at Re=15000 by hand

```python
def friction_factor(Re):
    """
    Darcy friction factor for smooth pipe.
    Re : Reynolds number [-]
    returns f [-]; laminar: 64/Re, turbulent: Blasius 0.316*Re^(-0.25)
    """
    if Re < 2100:
        return 64 / Re
    else:
        return 0.316 * Re ** (-0.25)

assert friction_factor(1800) == 64 / 1800
blasius_hand = 0.316 * (15000) ** (-0.25)
assert friction_factor(15000) == blasius_hand
print(f'f(1800) = {friction_factor(1800)}')
print(f'f(15000) = {friction_factor(15000)}, hand Blasius = {blasius_hand}')
print('both asserts passed')
```

### Problem 15: collected asserts

**Pseudocode:**
1. Redefine flow_regime, friction_factor, mixture_MW in one cell
2. Write 5 asserts (at least one per function; ≥2 with math.isclose + rel_tol)
3. All must pass on Restart & Run All

```python
def mixture_MW(y, MW):
    """
    Compute mixture molecular weight from mole fractions and component MWs.
    y  : dict of mole fractions [-]
    MW : dict of molecular weights [g/mol]
    returns mixture MW [g/mol]
    """
    M = 0.0
    for comp in y:
        M += y[comp] * MW[comp]
    return M

def flow_regime(rho, v, D, mu):
    """
    Classify pipe flow regime from Reynolds number.
    rho : density [kg/m3]
    v   : velocity [m/s]
    D   : diameter [m]
    mu  : viscosity [Pa.s = kg/(m.s)]
    returns (Re [-], regime [str])
    """
    Re = rho * v * D / mu
    if Re < 2100:
        regime = "laminar"
    elif Re <= 4000:
        regime = "transition"
    else:
        regime = "turbulent"
    return (Re, regime)

def friction_factor(Re):
    """
    Darcy friction factor for smooth pipe.
    Re : Reynolds number [-]
    returns f [-]; laminar: 64/Re, turbulent: Blasius 0.316*Re^(-0.25)
    """
    if Re < 2100:
        return 64 / Re
    else:
        return 0.316 * Re ** (-0.25)

# five asserts — at least one per function; ≥2 use math.isclose
assert mixture_MW({"N2": 1.0}, {"N2": 28.02}) == 28.02
assert math.isclose(mixture_MW({"A": 0.5, "B": 0.5}, {"A": 20.0, "B": 40.0}), 30.0, rel_tol=1e-12)
assert flow_regime(998, 0.1, 0.01, 0.001)[1] == "laminar"  # Re = 998
assert friction_factor(1800) == 64 / 1800
assert math.isclose(friction_factor(10000), 0.316 * 10000**(-0.25), rel_tol=1e-12)

print('all 5 asserts passed')
```

### Problem 16: multi-stream balance

**Pseudocode:**
1. Define FEED, DIST, BTMS dicts
2. For each component in FEED:
   - flow_out = DIST.get(comp, 0) + BTMS.get(comp, 0)
   - diff = flow_in - flow_out
3. Print reconciliation table
4. Assert |diff| < 1e-9 for each component and overall total

```python
FEED = {"benzene": 87.5, "toluene": 100.0, "xylene": 62.5} #kmol/h
DIST = {"benzene": 86.0, "toluene": 12.0}
BTMS = {"benzene": 1.5, "toluene": 88.0, "xylene": 62.5}

print(f'{"component":<12} {"in":>10} {"out":>10} {"diff":>12}')
total_in = 0.0
total_out = 0.0

for comp in FEED:
    flow_in = FEED[comp]
    flow_out = 0.0
    for stream in [DIST, BTMS]:
        flow_out += stream.get(comp, 0.0)
    diff = flow_in - flow_out
    print(f'{comp:<12} {flow_in:10.2f} {flow_out:10.2f} {diff:12.3e}')
    assert abs(diff) < 1e-9
    total_in += flow_in
    total_out += flow_out

assert abs(total_in - total_out) < 1e-9
print(f'{"TOTAL":<12} {total_in:10.2f} {total_out:10.2f} {total_in - total_out:12.3e}')
print('closure asserts passed')
```

### Problem 17: aliasing bug

**Pseudocode:**
1. Reproduce bug: case2 = base aliases the same list
2. Fix with .copy() and assert base intact
3. Fix again with slicing [:] and assert base intact

```python
# a) the bug: case2 = base makes both names point to the SAME list,
#    so mutating case2 also mutates base
base = [0.2, 0.3, 0.5]
case2 = base
case2[0] = 0.25
case2[1] = 0.25
print(f'buggy: base = {base} (got mutated via alias)')

# b) fix 1: .copy()
base = [0.2, 0.3, 0.5]
case2 = base.copy()
case2[0] = 0.25
case2[1] = 0.25
assert base == [0.2, 0.3, 0.5]
print(f'fix .copy(): base = {base}, case2 = {case2}')

# b) fix 2: slicing
base = [0.2, 0.3, 0.5]
case2 = base[:]
case2[0] = 0.25
case2[1] = 0.25
assert base == [0.2, 0.3, 0.5]
print(f'fix slicing: base = {base}, case2 = {case2}')
```

### Problem 18: grid-search optimum

**Pseudocode:**
1. Loop D from 0.02 to 0.20 step 0.005
2. Evaluate C(D) = 0.45e-4/D^5 + 120*D^2
3. Track D* and C* (minimum)
4. Confirm neighbours of D* give higher cost
5. Note grid spacing limits accuracy

```python
D = 0.02 #m
D_star = None
C_star = None
idx_star = None
grid = []

while D <= 0.20 + 1e-12:
    C = 0.45e-4 / D**5 + 120 * D**2
    grid.append((D, C))
    if C_star is None or C < C_star:
        C_star = C
        D_star = D
        idx_star = len(grid) - 1
    D += 0.005

print(f'D* = {D_star:.3f} m, C* = {C_star:.6f}')

# confirm interior optimum: both neighbours higher
assert grid[idx_star - 1][1] > C_star and grid[idx_star + 1][1] > C_star
print(f'neighbours: C({grid[idx_star-1][0]:.3f}) = {grid[idx_star-1][1]:.6f}, C({grid[idx_star+1][0]:.3f}) = {grid[idx_star+1][1]:.6f}')
print('interior optimum confirmed')
```

c) Grid search accuracy is limited by the spacing of the discrete grid — the true continuous optimum may lie between sampled points.

### Problem 19: normalize composition

**Pseudocode:**
1. Store raw mass fractions in a dict; compute raw sum
2. Normalize with dict comprehension: y_i / sum
3. Assert new sum ≈ 1 and ratios unchanged
4. Print before/after side by side

```python
w = {"CO2": 0.118, "H2O": 0.071, "N2": 0.742, "O2": 0.066} #- mass fractions (raw)

raw_sum = sum(w.values())
print(f'a) raw sum = {raw_sum}')

y = {comp: w[comp] / raw_sum for comp in w}
print(f'b) normalized sum = {sum(y.values())}')

assert abs(sum(y.values()) - 1.0) < 1e-12
# ratio of any two fractions unchanged
assert abs(y["CO2"] / y["N2"] - w["CO2"] / w["N2"]) < 1e-12
print('c) asserts passed (sum==1, ratios unchanged)')

print(f'{"comp":<6} {"before":>10} {"after":>10}')
for comp in w:
    print(f'{comp:<6} {w[comp]:10.6f} {y[comp]:10.6f}')
```
