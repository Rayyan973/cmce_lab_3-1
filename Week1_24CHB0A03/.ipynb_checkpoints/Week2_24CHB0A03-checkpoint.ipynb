### Problem 2: T_log

```python
T_log = [349.9, 350.4, 351.2, 350.8, 350.1] #K

#sum
total=0
for t in T_log:
    total += t
mean = total/len(T_log)

#max
maxi = -1
for t in T_log:
    if t > maxi:
        maxi = t

print(f'mean = {mean} : maximum= {maxi}')
```

```python
#normal unrestricted easier way to do this
print(f'mean = {sum(T_log)/len(T_log)} : maximum= {max(T_log)}')
```

### Problem 3: rho dict

```python
rho = {"water" : 998, "ethanol" : 789, "benzene" : 876}
print(rho)
rho["toluene"] = 867
print(rho)
rho["ethanol"] = 789.3
print(rho)

print(rho.get("acetone", -1.0))
```

here the last line returns -1 because the key "acetone" doesnt exist and hence defaults to the value passed (i.e. -1)
