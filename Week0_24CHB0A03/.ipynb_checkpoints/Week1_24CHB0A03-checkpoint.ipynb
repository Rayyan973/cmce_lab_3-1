### Problem 01: pressure

```python
P_atm = 2.5 #atm
P_Pa = P_atm * 101325.0 #Pascal
P_bar = P_atm * 1.01325 #bar
P_mmHg = P_atm * 760 #mmHg

print(f'P(atm) = {P_atm:.3f}, P(Pa) = {P_Pa:.2f}, P(bar) = {P_bar:.3f}, P(mmHg) = {P_mmHg:.0f}.')
```

### Problem 02: mole fractions

```python
M_methanol = 32.04 #g/mol
w_methanol = 40.0 #g
M_water = 18.015 #g/mol
w_water = 60.0 #g

mol_methanol = w_methanol / M_methanol
mol_water = w_water / M_water

total_moles = mol_methanol + mol_water

X_methanol = mol_methanol / total_moles
X_water = mol_water / total_moles

print(f'Mole fraction of Methanol = {X_methanol:.4f} : Mole fraction of Water = {X_water:.4f}')
print(f'Total = {X_methanol + X_water}')
```

### Problem 03: ideal gas

```python
T = 298 #K
P = 202650 #Pa
V = 0.5 #m^3
R = 8.314 #J/mol-K

n = (P*V) / (R*T) # 1Pa-m^3 = 1J so all set

print(f"n = {n:.3f} moles")
```

### Problem 09: antoine eqn

```python
A = 8.07131
B = 1730.63
C = 233.426

def antoine(A, B, C, T):
    """T is in C"""
    return 10 ** (A - (B / (C+T)))

temp = [25, 60, 100]
ans = [antoine(A, B, C, t) for t in temp] 

print(ans)

print(f'T={temp[0]}C; Psat = {ans[0]:.1f}mmHg')
print(f'T={temp[1]}C; Psat = {ans[1]:.1f}mmHg')
print(f'T={temp[2]}C; Psat = {ans[2]:.1f}mmHg')
```

##### Psat at 100C should be close to 1atm (760mmHg)

### Problem 10: reynolds number

```python
v = 1.2 #m/s
D = 52.5 #mm
rho = 998 #kg/m3
mu = 1.0e-3 #Pa-s

Re = D*v*rho/mu

flag = "Null"

if Re<2100:
    flag = "laminar"
elif 2100 <= Re <= 4000:
    flag = "in transition region"
else: 
    flag = "turbulent"

print(f'Re = {Re:.0f} -> {flag}')
```

### Problem 11: LMTD

```python
def lmtd(delT1, delT2):
    """temperatures can be anything"""
    if(delT1 <= 0 or delT2 <= 0):
        print("invalid data.")
        return

    if abs(delT1 - delT2) < 1e-9:
        return delT1
    else: return (delT1 - delT2) / math.log(delT1/delT2)

print(f'a) {lmtd(55, 25):.2f}, b) {lmtd(30, 30):.2f}')
```

### Problem 12: vanderwaal

```python
a = 0.3640 #Pa-m^6/mol^2
b = 4.267e-5 #m^3/mol
T = 350 #K
V = 1e-3 #m^3/mol

P_vdw = (R*T/(V-b)) - (a/(V**2))
P_ideal = R*T/V

print(f'a) VanDerWaals P = {P_vdw:.2f}; b) Ideal P = {P_ideal:.2f}; c) Percentage Deviation = {(P_ideal - P_vdw)*100/P_ideal:.2f}%')
```

### Problem 14: antoine

```python
def T_Antoine(A, B, C, P):
    
    return B/(A-math.log10(P)) - C

print(f'a) T = {T_Antoine(6.90565, 1211.033, 220.79, 400):.2f}C')
print(f'b) P at t=60.61 is: {antoine(6.90565, 1211.033, 220.79, 60.61):.2f}mmHg')
```

### Problem 15: dittus-boelter

```python
def nusselt(Re, Pr):
    """Re and Pr are both dimensionless"""
    return 0.023 * pow(Re, 0.8) * pow(Pr, 0.4)

Nu = nusselt(25000, 5.8)

print(f'a) Nu = {Nu:.1f} \nb) h = {(Nu * 0.62 / 0.04):.0f} W/m2-K')
```
