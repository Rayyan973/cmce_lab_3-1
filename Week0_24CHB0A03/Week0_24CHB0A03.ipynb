### Problem 09: antoine eqn

```python
A = 8.07131
B = 1730.63
C = 233.426

def antoine(A, B, C, T):
    return 10 ** (A - (B / (C+T)))

temp = [25, 60, 100]
ans = [antoine(A, B, C, t) for t in temp] 

print(ans)

print(f'T={temp[0]}C; Psat = {ans[0]:.1f}mmHg')
print(f'T={temp[1]}C; Psat = {ans[1]:.1f}mmHg')
print(f'T={temp[2]}C; Psat = {ans[2]:.1f}mmHg')
```

### Problem 10: reynolds number

```python
v = 1.2 #m/s
D = 52.5 #mm
rho = 998 #kg/m3
mu = 1.0e-3 #Pa-s

Re = D*v*rho/mu

flag = "Null"

if Re<2100:
    flag = "laminar"
elif 2100 <= Re <= 4000:
    flag = "in transition region"
else: 
    flag = "turbulent"

print(f'Re = {Re:.0f} -> {flag}')
```

### Problem 11: LMTD

```python
def lmtd(delT1, delT2):
    if(delT1 <= 0 or delT2 <= 0):
        print("invalid data.")
        return

    if abs(delT1 - delT2) < 1e-9:
        return delT1
    else: return (delT1 - delT2) / math.log(delT1/delT2)

print(f'a) {lmtd(55, 25):.2f}, b) {lmtd(30, 30):.2f}')
```

### Problem 12: vanderwaal

```python
a = 0.3640 #Pa-m^6/mol^2
b = 4.267e-5 #m^3/mol
T = 350 #K
V = 1e-3 #m^3/mol

P_vdw = (R*T/(V-b)) - (a/(V**2))
P_ideal = R*T/V

print(f'a) VanDerWaals P = {P_vdw:.2f}; b) Ideal P = {P_ideal:.2f}; c) Percentage Deviation = {(P_ideal - P_vdw)*100/P_ideal:.2f}%')
```

### Problem 17: reactor protection

```python
def reactor_action(T, P, L):
    if P > 12:
        return "TRIP: relief + feed cut"
    elif T > 450:
        return "TRIP: quench"
    elif L > 90 or L < 15:
        return "ALARM: level"
    elif T > 430 and P > 10:
        return "ALERT: approaching envelope"
    else:
        return "normal"

cases = [(455, 11, 50), (445, 12.5, 96), (435, 10.5, 40)]

for T, P, L in cases:
    print(f'(T={T} K, P={P} bar, L={L}%) -> {reactor_action(T, P, L)}')

# case 2: L=96% would trip ALARM, but P=12.5 > 12 fires first. priority matters hence
```

### Problem 18: Antoine bracket
Psat(40) < 400 < Psat(90) being True means the saturation temperature at 400 mmHg lies between 40 C and 90 C.

```python
A = 8.07131
B = 1730.63
C = 233.426
P_total = 400 #mmHg

Psat_40 = antoine(A, B, C, 40)
Psat_90 = antoine(A, B, C, 90)
bracketed = Psat_40 < P_total < Psat_90

print(f'Psat(40C) = {Psat_40:.1f} mmHg')
print(f'Psat(90C) = {Psat_90:.1f} mmHg')
print(f'bracketed = {bracketed}')
```

### Problem 19: ideal-gas calculator

```python
def ideal_gas_calc(T_in, P_in, V_in):
    T = float(T_in)
    P = float(P_in)
    V = float(V_in)

    if T <= 0:
        print("error: T is invalid (must be > 0)")
        return
    if P <= 0:
        print("error: P is invalid (must be > 0)")
        return
    if V <= 0:
        print("error: V is invalid (must be > 0)")
        return

    n = (P * V) / (R * T)
    print(f'n = {n:.3f} moles')

ideal_gas_calc("310", "150000", "0.25")
ideal_gas_calc("-40", "90000", "0.1")
```
